```python
import scanpy as sc
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import pandas as pd
```

```python
adata = sc.read_h5ad('xxx/PaaSc_Data/MouseBrain/adata_benchmark.h5ad')
adata = sc.AnnData(X=adata.X, obs=adata.obs, obsm=adata.obsm)

structure_Top200marker_score = pd.read_csv('xxx/PaaSc_Data/MouseBrain/Visium_Mousebrain_cellmarker_top200_PaaSc_score.tsv', sep='\t', index_col=0)
structure_Top200marker_score = structure_Top200marker_score.loc[adata.obs.index]
adata.obs = adata.obs.join(structure_Top200marker_score)
```

```python
markers=['L6.CT', 'L6b', 'L6.IT', 'L2.3.IT', 'L5.PT',  'L5.IT','L4', 'Oligo']

cmap = LinearSegmentedColormap.from_list(
    "my_cmap", ["blue", "yellow", "red"], N=256
)

sc.pl.spatial(adata, color=markers, spot_size=130,  color_map=cmap, show=False)

plt.savefig("Fig6.pdf", format="pdf", dpi=300)
```

```python
sc.pl.spatial(adata, color='VLMC', vmax=4,spot_size=130,  color_map=cmap, show=False)
plt.savefig("Fig6_VLMC.pdf", format="pdf", dpi=300)
sc.pl.spatial(adata, color='Vip', spot_size=130,  color_map=cmap, show=False)
plt.savefig("Fig6_Vip.pdf", format="pdf", dpi=300)
```
